```python
########## DATA ##########


path_img1 = r"C:\Users\guigu\Documents\pro_asus\vigisar\data\preprocessed\gironde\gironde_pre.tif"
path_img2 = r"C:\Users\guigu\Documents\pro_asus\vigisar\data\preprocessed\gironde\gironde_post.tif"
```

```python
########## MAIN_dtod Execution ##########

# Execution parameters
n_tiles = 1        # Number of tiles (n x n)
k_factor = 1.15     # Otsu scaling factor
win_size = 30      # Density window size
density_thr = 0.5  # Density threshold
min_area = 3000    # Minimum component size
output_path = None # Saving the output mask is optional, set to None to skip saving 

# Run the main_dtod detection algorithm
profile, change_mask = main_dtod.main_dtod(
    path_img1, 
    path_img2, 
    n=n_tiles, 
    k=k_factor, 
    p=win_size, 
    d=density_thr, 
    a=min_area, 
    out_path=None
)

print(f"Mask computed. Shape: {change_mask.shape}")

def plot_results(path_img1, path_img2, mask):
    """
    Display the two input images and the resulting change mask with pixel coordinates.
    """
    with rasterio.open(path_img1) as src1, rasterio.open(path_img2) as src2:
        # Read first band for visualization
        img1 = src1.read(1)
        img2 = src2.read(1)
        
    fig, ax = plt.subplots(1, 3, figsize=(18, 7)) # Slightly taller for axis labels

    # Configuration for all subplots
    titles = ["Image Date 1 (S1)", "Image Date 2 (S1)", "Detected Change Mask"]
    data = [np.abs(img1), np.abs(img2), mask]
    cmaps = ['gray', 'gray', 'hot']

    for i in range(3):
        ax[i].imshow(data[i], cmap=cmaps[i])
        ax[i].set_title(titles[i])
        
        # Enable axis to see pixel counts
        ax[i].set_xlabel("Width (pixels)")
        ax[i].set_ylabel("Height (pixels)")
        ax[i].grid(False) # Ensure grid doesn't overlap the image

    plt.tight_layout()
    plt.show()

# Run visualization
plot_results(path_img1, path_img2, change_mask)
```
